```python
import pandas as pd
import os
```

```python
path = "./fg/"
file_name = []
for root, dirs, files in os.walk(path):
    for name in files:
        file_name.append(name)
df = pd.DataFrame(file_name, columns=['file_name'])
```

```python
df.sample()
```

```python
pp = path + df.sample().file_name.values[0]
pp
```

```python
tt = pd.DataFrame()
for index, row in df.iterrows():
    file_name = row.file_name
    try:
        pp = path + file_name
        tb = pd.read_pickle(pp)
        description = tb.head(1)['zhengcewenjian.description'].values[0]
        aging = tb.head(1)['zhengcewenjian.aging'].values[0]
        title = tb.head(1).title.values[0]
        content = tb.head(1).content.values[0]
        publishedTimeStr = tb.head(1).publishedTimeStr.values[0]

        dic = {'file_name':file_name, 'description':description, 'aging':aging, 'title':title, 'content':content, 'time':publishedTimeStr.split(' ')[0]}
        tt = pd.concat([tt, pd.DataFrame(dic, index=[0])], ignore_index=True)
    except:
        print(file_name)
        continue
    
```

```python
tt.sample(10)
```

```python
tt.aging = tt.aging.astype(str)
tt.time = tt.time.astype(str)
```

```python
tt.aging.value_counts()
```

```python
tb = tt[~tt.aging.str.contains('尚未生效|全文失效|已修改|全文废止')].copy()
tb = tb.loc[:, ['title', 'time', 'content']]
```

```python
tb.sample(10)
```

```python
for index, row in tb.iterrows():
    title = row.title
    if '关于' in title:
        title = title.split('关于')[-1]
    tb.loc[index, 'title'] = title
```

```python
tb.to_pickle('./tax_law.pkl')
```
