```python
import pandas as pd
import numpy as np
import time as tiime
import random
import os
import string 
import string
import urllib.request
import time
import re
import json
import requests
from tqdm import tqdm
import math
```

```python
tb = pd.DataFrame()

for i in tqdm(range(2)):
    url = "https://www.chinatax.gov.cn/getFileListByCodeId"
    headers = {"Accept": "*/*","Content-Type": "application/x-www-form-urlencoded; charset=UTF-8","Origin": "https://fgk.chinatax.gov.cn","Referer": "https://fgk.chinatax.gov.cn/","User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:142.0) Gecko/20100101 Firefox/142.0",}
    sent_form = {"codeId": "","channelId": "29a88b67e4b149cfa9fac7919dfb08a5","page": str(i),"size": "10"}
    res = requests.post(url, headers=headers, data=sent_form, timeout=(10, 20))
    if res.status_code == 200:
        d0 = res.json()
        d1 = d0['results']['data']['results']
        for data in d1:
            record = {"title": data.get("title"), "url": data.get("url"), "publishedTimeStr": data.get("publishedTimeStr"), "channelName": data.get("channelName")}
            for domain in data.get("domainMetaList", []):
                for item in domain.get("resultList", []):
                    key = item.get("key")
                    val = item.get("value")
                    record[key] = val
            df = pd.DataFrame([record])
            tb = pd.concat([tb, df], ignore_index=True)
        time.sleep(random.uniform(20, 30))
    else:
        print(i)
        time.sleep(random.uniform(20, 30))
```

```python
import requests

# Define the URL you want to scrape
url = 'https://sichuan.chinatax.gov.cn/col/col320/index.html'

# Set up the headers as seen in the request headers
headers = {
    'Accept': 'text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8',
    'Accept-Encoding': 'gzip, deflate, br, zstd',
    'Accept-Language': 'en-US,en;q=0.5',
    'Connection': 'keep-alive',
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:145.0) Gecko/20100101 Firefox/145.0',
    'Host': 'sichuan.chinatax.gov.cn',
    'Referer': 'https://sichuan.chinatax.gov.cn/col/col320/index.html?uid=864&pageNum=2',
    'Upgrade-Insecure-Requests': '1',
}

# Send GET request with headers, disabling SSL verification
response = requests.get(url, headers=headers, verify=False)

# Check if the request was successful
if response.status_code == 200:
    # If successful, parse the content using BeautifulSoup
    page_content = response.text
    print(page_content)
else:
    print(f"Failed to retrieve the page. Status code: {response.status_code}")
```

```python
import requests
from bs4 import BeautifulSoup

# URL for the "政策法规库" page
url = "http://sichuan.chinatax.gov.cn/col/col19973/index.html"

# Send a GET request to the page
response = requests.get(url)
response.encoding = 'utf-8'  # Ensure correct encoding for Chinese content

# Parse the HTML content using BeautifulSoup
soup = BeautifulSoup(response.text, 'html.parser')

# Find all the policy links
policy_links = soup.find_all('a', href=True, title=True)

# Filter links that point to actual policies (those with valid title)
for link in policy_links:
    policy_title = link.get('title')
    policy_url = link.get('href')

    # Skip links with no valid title or broken links
    if not policy_title or "javascript" in policy_url:
        continue

    # Construct the full URL for each policy (in case the link is relative)
    full_policy_url = "http://sichuan.chinatax.gov.cn" + policy_url

    # Send a request to get the policy content
    policy_response = requests.get(full_policy_url)
    policy_soup = BeautifulSoup(policy_response.text, 'html.parser')

    # Extract the policy text (adjust based on page structure)
    policy_content = policy_soup.find('div', class_='TRS_Editor')  # The content area may vary based on actual HTML

    if policy_content:
        print(f"Title: {policy_title}")
        print(f"URL: {full_policy_url}")
        print(f"Content: {policy_content.get_text(strip=True)}\n")
    else:
        print(f"Failed to extract content from: {full_policy_url}")
```
