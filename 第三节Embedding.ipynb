```python
# Requires transformers>=4.51.0
# Requires sentence-transformers>=2.7.0
from sentence_transformers import SentenceTransformer
```

```python
# pip install modelscope
# D:
# mkdir model
# cd model
# mkdir Qwen3-Embedding-0.6B
# modelscope download --model Qwen/Qwen3-Embedding-0.6B --local_dir ./Qwen3-Embedding-0.6B

model = SentenceTransformer("D:\\model\\Qwen3-Embedding-0.6B")
```

```python
# The queries and documents to embed
queries = [
    "2024年度万泰生物公司营业收入为224,540.32万元，2023年度万泰生物公司营业收入为551,078.25万元。营业收入为万泰生物公司合并利润表重要组成项目，真实性及截止性可能存在潜在错报。因此我们将营业收入确认识别为关键审计事项。",
]
documents = [
    '2019年度万泰生物公司营业收入为118,375.42万元，2020年度万泰生物公司营业收入为235,425.68万元，为万泰生物公司合并利润表重要组成项目，营业收入的真实性及截止性可能存在潜在错报。因此我们将营业收入确认识别为关键审计事项。',
    '2020年度万泰生物公司营业收入为235,425.68万元，2021年度万泰生物公司营业收入为575,032.91万元，为万泰生物公司合并利润表重要组成项目，营业收入的真实性及截止性可能存在潜在错报。因此我们将营业收入确认识别为关键审计事项。',
    '2021年度万泰生物公司营业收入为575,032.91万元，2022年度万泰生物公司营业收入为1,118,518.87万元，为万泰生物公司合并利润表重要组成项目，营业收入的真实性及截止性可能存在潜在错报。因此我们将营业收入确认识别为关键审计事项。',
    '2022年度万泰生物公司营业收入为1,118,518.87万元，2023年度万泰生物公司营业收入为551,078.25万元，为万泰生物公司合并利润表重要组成项目，营业收入的真实性及截止性可能存在潜在错报。因此我们将营业收入确认识别为关键审计事项。',
]
```

```python
query_embeddings = model.encode(queries, prompt_name="query")
document_embeddings = model.encode(documents)
similarity = model.similarity(query_embeddings, document_embeddings)
print(similarity)
```
